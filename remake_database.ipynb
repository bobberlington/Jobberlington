# Recreate Database

This just allows you to reset the database or create it if it doesn't exist.

```python
import sqlite3
con = sqlite3.connect("jobs.db")
cur = con.cursor()
try:
    cur.execute("""DROP TABLE jobs""") 
except:
    pass
cur.execute("""CREATE TABLE jobs(
                            id integer PRIMARY KEY AUTOINCREMENT,
                            score integer,
                            date datetime,
                            company varchar,
                            title varchar,
                            url varchar,
                            details varchar,
                            description varchar,
                            analysis varchar,
                            logo varchar,
                            description_html varchar,
                            confidence_rating varchar,
                            job_summary varchar,
                        );""")
con.commit()
```
